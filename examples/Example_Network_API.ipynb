# Example Network API
This example demonstrates how to use the network API of Pupil Core to receive gesture and object detection events.

## Requirements
Connect the tracker and start the Pupil Core. The Pupil Core should be running the plugins for object detection and gaze gesture detection. To check if they are enabled, go to the Pupil Core GUI and click on the 'Plugin Manager' tab. The plugins should be listed there.

## Setup
First, we need to import the necessary libraries and request the sub and pub ports.

```python
import zmq
import msgpack
import pprint
```

```python
ctx = zmq.Context()
socket = ctx.socket(zmq.REQ)

# Define the IP and port of the Pupil Core
ip = 'localhost'
port = 50020

socket.connect(f'tcp://{ip}:{port}')

# Request 'SUB_PORT' for reading data
socket.send_string('SUB_PORT')
sub_port = socket.recv_string()

# Request 'PUB_PORT' for writing data
socket.send_string('PUB_PORT')
pub_port = socket.recv_string()

socket.close()
```

```python
def create_socket(ctx_c, ip_c, topics):
    '''
    Create a socket to subscribe to the specified topics.
    :param ctx_c:
    :param ip_c:
    :param topics:
    :return:
    '''
    sub = ctx_c.socket(zmq.SUB)
    sub.connect(f'tcp://{ip_c}:{sub_port}')
    for topic in topics:
        sub.subscribe(topic)
    return sub
```

## Receiving Object Detection Events

```python
socket_sub = create_socket(ctx, ip, ['blink_gestures'])
for i in range(10):
    topic = socket_sub.recv_string()
    payload = socket_sub.recv()
    msg = msgpack.unpackb(payload, raw=False)
    pprint.pprint(msg)
    
socket_sub.close()
```

## Receiving Gesture Events

```python
socket_sub = create_socket(ctx, ip, ['gestures'])
for i in range(10):
    topic = socket_sub.recv_string()
    payload = socket_sub.recv()
    msg = msgpack.unpackb(payload, raw=False)
    pprint.pprint(msg)
    
socket_sub.close()
```
